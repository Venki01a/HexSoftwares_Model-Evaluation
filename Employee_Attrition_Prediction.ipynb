# Distribution Of Employee Attrition

```python
colors = {"No": "#729ECE", "Yes": "#E15759"}

sns.set_theme(style = "whitegrid")
plt.figure(figsize = (9,6))

ax = sns.countplot(x = "Attrition",palette = colors, data = data)
total = len(data)
for p in ax.patches:
    count = p.get_height()
    x = p.get_x() + p.get_width() / 2
    percentage = f'({100 * count / total:.1f}%)'
    ax.annotate(f'{count:.0f}', (x, count), 
                ha='center', va='center', 
                size=14, 
                weight='bold', 
                xytext=(0, 20), 
                textcoords='offset points')
    
    ax.annotate(percentage, (x, count), 
                ha='center', va='center', 
                size=11, color='gray', 
                xytext=(0, 5), 
                textcoords='offset points')
    
plt.ylim(0,1400)
plt.title("Distribution of Employee Attrition",fontsize = 16,pad = 20)
plt.xlabel("Employee Attrition Status",fontsize = 12)
plt.ylabel("Number of Employees",fontsize = 12)
plt.show()
```

An examination of our target variable, 'Attrition,' shows that the vast majority of the 1,470 employees (83.9%, or 1,233 individuals) remained with the company. The remaining 16.1% (237 individuals) represent the minority group who left their jobs.

This distribution is a key finding that forms the basis of our analysis. Using visualizations, we now aim to identify the key characteristics that distinguish this minority group from the majority and uncover the factors that trigger the decision to leave.

# Attrition Rate By Department

```python
avg = (data['Attrition'] == 'Yes').groupby(data['Department']).mean() * 100

print(avg.map('{:.2f}%'.format))
```

```python
from matplotlib.ticker import PercentFormatter
data['Attrition_numeric'] = data['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

order = data.groupby('Department')['Attrition_numeric'].mean().sort_values(ascending=False).index

ax = sns.barplot(
    data=data,
    x="Department",
    y="Attrition_numeric", 
    order=order,
    errorbar = None,
    palette="coolwarm"
)

ax.yaxis.set_major_formatter(PercentFormatter(1.0))

for p in ax.patches:
    percentage = f'{p.get_height() * 100:.1f}%'
    ax.annotate(percentage,
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom',
                size=16,
                xytext=(0, 5),
                textcoords='offset points')

plt.title("Attrition Rate by Department", fontsize=16, pad=20)
plt.xlabel("Department", fontsize=12)
plt.ylabel("Attrition Rate (%)", fontsize=12)

plt.ylim(0, data.groupby('Department')['Attrition_numeric'].mean().max() * 1.2)

plt.show()
```

When examining departmental turnover rates, the Sales department has the highest rate at 20.6%. Human Resources follows at 19%, while the Research & Development department has a significantly lower rate at 13.8%.

This finding is particularly striking when viewed alongside the previous "Average Salary by Department" chart. Contrary to expectations, Sales, which has the highest average salary, also leads in turnover rates.

This paradox strongly demonstrates that salary alone is not a determining factor in employee retention. It suggests that, particularly in fast-paced departments like Sales, factors such as job stress, target pressure, and work-life balance may outweigh salary.

The chart reveals a clear distinction between the roles: roles like Sales Representatives, Laboratory Technicians, and Human Resources pose a high risk with rates exceeding 20%, while senior and managerial roles like Directors and Managers exhibit a relatively stable pattern with rates below 7%.

This finding also illuminates the "salary-turnover paradox" discussed earlier. The high turnover in the Sales department appears to be largely driven by the entry-level Sales Representative role. This suggests that employees in this role, in particular, face significant challenges in the company's aging process.

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['Attrition']=le.fit_transform(data['Attrition'])
```

# Label Encoding Using get dummies

```python
data = pd.get_dummies(data, columns=categoric.columns.drop(["Attrition"]), drop_first=True)
```

# Scaling the Numeric Columns

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_columns = ['Age', 'MonthlyIncome', 'YearsAtCompany',
                  "DailyRate","HourlyRate","MonthlyRate",
                  "PercentSalaryHike","DistanceFromHome",
                  "Education","EnvironmentSatisfaction",
                  "NumCompaniesWorked","TotalWorkingYears",
                  "YearsInCurrentRole","YearsSinceLastPromotion"]
data[scaled_columns] = scaler.fit_transform(data[scaled_columns])
```

```python
data
```

# Create X and Y variable for training model

```python
x = data.drop(columns=['Attrition'])
y = data['Attrition']
```

# Train - Test Split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Create Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr
```

```python
lr.fit(x_train,y_train)
```

# Make Predictions Using Logistic Regression Model

```python
lr_pred=lr.predict(x_test)
lr_pred
```

# Accuracy From Logistic Regression Model

```python
from sklearn.metrics import accuracy_score
lr_accuracy = accuracy_score(y_test, lr_pred)
print(f"Logistic Regression Model Accuracy: {lr_accuracy * 100:.2f}%")
```

# Precision

```python
from sklearn.metrics import precision_score
lr_precision = precision_score(y_test, lr_pred)
print(f"Logistic Regression Model Precision: {lr_precision * 100:.2f}%")
```

# Recall


```python
from sklearn.metrics import recall_score
lr_recall = recall_score(y_test, lr_pred)
print(f"Logistic Regression Model Recall: {lr_recall * 100:.2f}%")
```

# F-1 Score

```python
from sklearn.metrics import  f1_score
lr_f1 = f1_score(y_test, lr_pred)
print(f"Logistic Regression Model F1 Score: {lr_f1 * 100:.2f}%")
```

# Classification Report

```python
from sklearn.metrics import classification_report
classReport = classification_report(y_test, lr_pred)
print(classReport)
```

# Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, lr_pred)
cm
```

# Conclusion

This analysis aimed to uncover the dynamics behind employee attrition using the IBM HR dataset. Through our exploratory data analysis, the key factors influencing an employee's decision to leave have been identified, and a general "risk profile" has clearly emerged.

## Key Findings from the Project:
   ###  Overtime is the Biggest Red Flag: 
   The most striking finding from our analysis is the impact of overtime on attrition. An employee who works overtime is approximately three      times more likely to leave than one who does not. This indicates that the company's overtime culture and workload distribution are the most    critical issues concerning employee retention.

### Dissatisfaction Triggers Resignation:
   Both the Job Satisfaction and Environment Satisfaction analyses told the same clear story: the group with the highest concentration of         employees who left are those with "Low" satisfaction levels. A poor job experience or a negative work environment are among the strongest      drivers pushing employees to resign.

### Career Stagnation is a Major Push Factor:
   Analyses of promotions and career paths revealed the company's struggle to retain employees in specific roles. The "Sales Representative"      role is particularly alarming, with an attrition rate approaching 40%. Generally, when an employee's waiting time for a promotion extends      for many years beyond the norm, it becomes a powerful reason to leave. Employees who feel "forgotten" lose their loyalty to the company.

### The Salary Paradox: Money Isn't Everything:
   Our analysis proves that while salary is an important factor, it is not sufficient on its own. The clearest evidence of this is that the       Sales department, which has the highest average salary, also has one of the highest attrition rates. This shows that in high-stress,           target- driven roles, even a good salary cannot compensate for other negative factors.

## Conclusion: Who is Leaving?
Combining all these findings, the profile of a "typical" employee who leaves this company is as follows:

An experienced professional who is dissatisfied with their job or environment, sees no path for career progression, and is very likely working overtime.

This analysis strongly suggests that for the company to improve its employee retention, its strategies must focus not only on salary increases but also on balancing workloads, improving the work environment, and providing clear career paths.
